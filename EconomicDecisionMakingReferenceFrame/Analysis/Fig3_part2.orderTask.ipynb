```python
def importAndPreprocess(dirPath,activityFileName):
    import os
    import sys

    
    with np.load(os.path.join(dirPath,activityFileName),allow_pickle=True) as f:
        x = f['x']
        trial_params = f['trial_params']
        model_output = f['model_output']
        model_state = f['model_state']
        mask = f.get('mask', None)
    
    if mask is None:
        temp = np.mean(model_output[:,300:,:],1)
    else: 
        temp = np.mean(mask * model_output,1)
    choiceLR = temp[:,1]>temp[:,0]
    choiceLR = choiceLR*2-1 # pos right high, neg left high
    
    choiceFrame = [trial_params[i]['choiceFrame'] for i in range(len(trial_params))]
    
    
    locAB = [(1 if trial_params[i]['locAB']=='AB' or trial_params[i]['locAB']=='12' else -1) for i in range(len(trial_params))]
    loc12 = locAB
    seqAB = [(1 if trial_params[i]['seqAB']=='AB' else -1) for i in range(len(trial_params))]
    
    choiceAB = np.array([(choiceLR[i] * locAB[i] if choiceFrame[i]=='juice' else choiceLR[i] * locAB[i] * seqAB[i]) for i in range(len(trial_params)) ])  # pos B neg A
    choice12 = np.array([(choiceAB[i] * seqAB[i] if choiceFrame[i]=='juice' else choiceLR[i] * locAB[i]) for i in range(len(trial_params)) ])  # pos 2 neg 1
    choiceAB = np.array(['B' if choiceAB[i]>0 else 'A' for i in range(len(trial_params))])
    choice12 = np.array(['2' if choice12[i]>0 else '1' for i in range(len(trial_params))])
    
    qAs = np.array([trial_params[i]['qA'] for i in range(len(trial_params))])
    qBs = np.array([trial_params[i]['qB'] for i in range(len(trial_params))])
    seqAB = np.array([trial_params[i]['seqAB']for i in range(len(trial_params))])

    return x,trial_params,model_state,choice12,choiceAB,choiceLR,qAs,qBs,seqAB
```

```python
import re
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA

def find_analysis_dir():
    candidates = [
        Path.cwd(),
        Path.cwd() / 'Analysis',
        Path.cwd() / 'EconomicDecisionMakingReferenceFrame' / 'Analysis',
    ]
    for candidate in candidates:
        if (candidate / 'Fig3_part2.orderTask.ipynb').exists():
            return candidate.resolve()
    raise FileNotFoundError('Cannot locate the EconomicDecisionMakingReferenceFrame/Analysis directory.')

def select_model_dir(task_dir, ensemble_index=0):
    pattern = re.compile(rf'_{ensemble_index}_(?:Fail)?$')
    candidates = [path for path in task_dir.iterdir() if path.is_dir() and pattern.search(path.name)]
    if not candidates:
        raise FileNotFoundError(f'No ensemble {ensemble_index} directory found under {task_dir}.')
    return sorted(candidates, key=lambda path: ('Fail' in path.name, path.name))[0]

def load_activity(model_dir, filename):
    with np.load(model_dir / filename, allow_pickle=True) as data:
        model_state = data['model_state']
        trial_params = data['trial_params']
    qAs = np.asarray([params['qA'] for params in trial_params], dtype=float)
    qBs = np.asarray([params['qB'] for params in trial_params], dtype=float)
    seqAB = np.asarray([params['seqAB'] for params in trial_params])
    return model_state, qAs, qBs, seqAB

def projected_fixation_field(model_dir, pca_obj, limits):
    xmin, xmax, ymin, ymax = limits
    with np.load(model_dir / 'weightFinal.npz', allow_pickle=True) as weights:
        W_rec = weights['W_rec']
        W_in = weights['W_in']
        b_rec = weights['b_rec']

    def state_derivative(states, task_input):
        states_t = states.T
        recurrent = W_rec @ np.maximum(states_t, 0) + b_rec[:, None]
        input_drive = (W_in @ task_input)[:, None]
        return ((-states_t + recurrent + input_drive) / 100).T

    basis = pca_obj.components_[:2]
    pc1, pc2 = basis
    xv, yv = np.meshgrid(np.arange(xmin, xmax, 2), np.arange(ymin, ymax, 2))
    state_grid = np.outer(xv.ravel(), pc1) + np.outer(yv.ravel(), pc2) + pca_obj.mean_
    fixation_input = np.zeros(W_in.shape[1])
    fixation_input[-1] = 1
    vectors = state_derivative(state_grid, fixation_input) @ basis.T
    vectors = vectors.reshape((*xv.shape, 2))
    return xv, yv, vectors

analysis_dir = find_analysis_dir()
saved_local = analysis_dir.parent / 'Training' / 'savedForLocal'
figure_dir = analysis_dir / 'Figure'
figure_dir.mkdir(parents=True, exist_ok=True)
model_specs = {
    'combined': saved_local / 'spatialTask_combined',
    'seperated': saved_local / 'spatialTask_seperated',
}
ensemble_index = 0
limits = (-16, 22, -17, 14)
spatial_plot_context = {}

for model_name, task_dir in model_specs.items():
    model_dir = select_model_dir(task_dir, ensemble_index)
    model_state, qAs, qBs, seqAB = load_activity(model_dir, 'activitityTest.npz')
    K, T, N = model_state.shape
    pca_obj = PCA(n_components=4)
    pca_obj.fit(model_state[:, 50:250, :].reshape(K * 200, N))

    fig = plt.figure(figsize=(4, 3), dpi=150, constrained_layout=True)
    ax = fig.add_subplot()
    oranges = plt.get_cmap('Oranges', lut=5)
    for color_index, qA in enumerate([0, 1, 2, 3, 4]):
        condition = np.isclose(qAs, qA) & (seqAB == 'AB')
        if not np.any(condition):
            raise ValueError(f'{model_name}: no phase-1 trials found for qA={qA}.')
        trace = pca_obj.transform(model_state[condition, 50:151, :].mean(axis=0))
        trace_markers = trace[np.arange(10, 110, 10), :]
        ax.plot(trace[:, 0], trace[:, 1], color=oranges(color_index), label=f'{qA}A')
        ax.plot(trace_markers[:, 0], trace_markers[:, 1], marker='.', linestyle='', color=oranges(color_index))

    xv, yv, vectors = projected_fixation_field(model_dir, pca_obj, limits)
    ax.quiver(xv, yv, vectors[:, :, 0], vectors[:, :, 1], label='__projected derivatives', color='grey')
    ax.set_xlabel('PC1')
    ax.set_ylabel('PC2')
    ax.set_aspect('equal', 'box')
    ax.set_xlim(limits[0], limits[1])
    ax.set_ylim(limits[2], limits[3])
    ax.legend(loc='lower right', fontsize=8, frameon=True)
    output_path = figure_dir / f'Fig5E_{model_name}.pdf'
    fig.savefig(output_path, bbox_inches='tight')
    plt.show()
    plt.close(fig)
    spatial_plot_context[model_name] = {'model_dir': model_dir, 'pca': pca_obj, 'field': (xv, yv, vectors)}
    print(f'Saved: {output_path}')
```

```python
phase2_qA = [2.0, 4.0]
phase2_qB = [1.7, 3.4, 6.8]
phase2_color_indices = [2, 4, 8]

for model_name, context in spatial_plot_context.items():
    model_dir = context['model_dir']
    pca_obj = context['pca']
    xv, yv, vectors = context['field']
    model_state, qAs, qBs, seqAB = load_activity(model_dir, 'activitityTestGrid.npz')

    fig = plt.figure(figsize=(4, 3), dpi=150, constrained_layout=True)
    ax = fig.add_subplot()
    blues = plt.get_cmap('Blues', lut=9)

    for qA, linestyle in zip(phase2_qA, ['--', '-']):
        for qB, color_index in zip(phase2_qB, phase2_color_indices):
            condition = np.isclose(qAs, qA) & np.isclose(qBs, qB) & (seqAB == 'AB')
            if not np.any(condition):
                raise ValueError(f'{model_name}: no phase-2 trials found for qA={qA}, qB={qB}.')
            trace = pca_obj.transform(model_state[condition, 150:251, :].mean(axis=0))
            color = blues(color_index)
            ax.plot(trace[:, 0], trace[:, 1], color=color, linestyle=linestyle, label=f'{qA:g}A:{qB:g}B')
            ax.plot(trace[0, 0], trace[0, 1], color=color, marker='o', markerfacecolor='none')
            ax.plot(trace[-1, 0], trace[-1, 1], color=color, marker='o')

    ax.quiver(xv, yv, vectors[:, :, 0], vectors[:, :, 1], label='__projected derivatives', color='grey')
    ax.set_xlabel('PC1')
    ax.set_ylabel('PC2')
    ax.set_aspect('equal', 'box')
    ax.set_xlim(limits[0], limits[1])
    ax.set_ylim(limits[2], limits[3])
    ax.legend(loc='best', fontsize=7, frameon=True)
    output_path = figure_dir / f'Fig5E_phase2_withLegend_{model_name}.pdf'
    fig.savefig(output_path, bbox_inches='tight')
    plt.show()
    plt.close(fig)
    print(f'Saved: {output_path}')
```
